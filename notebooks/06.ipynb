```python
from dotenv import load_dotenv
load_dotenv
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a:int , b:int):
    """
       it will return the sum of two number

       Args:
         a : Number One 
         b : Number Two
    """

    return a+b 




@tool
def mul_numbers(a:int , b:int):
    """
       it will return the mul of two number

       Args:
         a : Number One 
         b : Number Two
    """

    return a*b 
```

```python
add_numbers.invoke({"a":12 ,"b" : 13})
```

```python
from langchain_groq import ChatGroq

from langchain.agents import create_agent
llm = ChatGroq(model="openai/gpt-oss-20b")
```

```python
agent = create_agent(
    model = llm ,
    tools = [add_numbers , mul_numbers],
    system_prompt="You are a math teacher and use only tools for calc"
)
```

```python
res = agent.invoke({"messages":[{"role" : "user" , "content" : "What is 2*5+10 ?"}]})
print(res["messages"][-1].content)
```
