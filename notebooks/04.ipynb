```python
from langchain_groq import ChatGroq

llm = ChatGroq(model="openai/gpt-oss-20b")
```

```python
text = "Hola! I am cristino ronaldo !!!" \
"Im the mister champions leaguge ...." \
"my age is 49"
```

```python
res = llm.invoke(f"what is my name and what is my age from the text : {text}" )
res.content
```

```python
from pydantic import BaseModel , Field 
from typing import List

class ResponseStructure(BaseModel):
    name : str = Field(description = "complete name")
    age : int= Field(description = "age")

structured_llm = llm.with_structured_output(ResponseStructure)
```

```python
res = llm.invoke(f"what is my name and what is my age from the text : {text}" )
res.model_dump_json()
```

```python
from typing import List
from pydantic import BaseModel, Field
class Movies(BaseModel):
    title:str = Field(description="Movie Title")
    year:int = Field(description="Movie Released year")

class AllMovies(BaseModel):
    movies : List[Movies]

movie_llm = llm.with_structured_output(AllMovies)
```

```python
movie_llm.invoke("Give three me one Sci-Fi movie")
```
