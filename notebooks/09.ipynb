#### Document text loader


```python
from langchain_community.document_loaders import TextLoader , PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader = TextLoader("../data/speech.txt")
texts = loader.load()
```

```python
pdfloader = PyPDFLoader("../data/resume1 (1).pdf")
pdfs = pdfloader.load()
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size =25 , chunk_overlap = 20)
```

```python
text_chunks = splitter.split_documents(texts)
```

```python
text_chunks
```

```python
pdf_chunks = splitter.split_documents(pdfs)
len(pdf_chunks)
```
